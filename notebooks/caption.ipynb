```python
import dotenv
import os

dotenv.load_dotenv()
gemini_api_key = os.environ["GEMINI_API_TOKEN"]
```

```python
from google import genai
```

```python
client = genai.Client(api_key=gemini_api_key)
```

```python
import PIL

img_path = "data/things-eeg2/imgs/training_images/00001_aardvark/aardvark_01b.jpg"
with open(img_path, "rb") as f:
    img_bytes = f.read()
    

```

```python
from pathlib import Path


requests = []
for img_path in list((Path("data") / "things-eeg2" / "imgs").rglob("*.jpg"))[:3]:
    file = client.files.upload(file=img_path)

    req = {
        "key": img_path,
        "request": {
            "contents": {
                "parts": [
                    {
                        "fileData": file.name
                    }
                ]
            }
        }
    }
    requests.append(req)
```

```python
requests
```

```python
response = client.models.generate_content(
    model='gemini-2.5-flash',
    config=genai.types.GenerateContentConfig(
        system_instruction='You are given an image. Generate a single, detailed caption that can be used directly as a prompt for a text-to-image model such as Stable Diffusion to faithfully recreate the scene. The caption must be visually precise, descriptive, and neutral. Do not assume any prior context. Avoid storytelling, interpretation, or emotional language. Describe only what is clearly visible.'
    ),
    contents=[
      genai.types.Part.from_bytes(
        data=img_bytes,
        mime_type='image/jpeg',
      ),
    ]
  )

print(response.text)
```
